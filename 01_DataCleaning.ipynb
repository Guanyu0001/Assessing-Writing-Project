```python
# ======================================================
# Analysis Pattern Code 01 — Data Cleaning
# ======================================================

import pandas as pd

# Load dataset (downloaded manually from Kaggle)
df = pd.read_csv("./data/persuade_2.0_human_scores_demo_id_github.csv")

# Filter: Independent task and Distance learning prompt
df = df[(df["task"] == "Text dependent") & (df["prompt_name"] == "Facial action coding system")]

# Keep only needed columns
df = df[["essay_id_comp", "full_text", "gender","race_ethnicity", "holistic_essay_score"]]

# Collapse holistic_essay_score from 1–6 to 1–3
df["holistic_essay_score_3cat"] = df["holistic_essay_score"].map(
    {1: 1, 2: 1, 3: 2, 4: 2, 5: 3, 6: 3}
)

# Save cleaned dataset
df.to_csv("./data/persuade_clean.csv", index=False)

# Preview
print(df.head())
```
