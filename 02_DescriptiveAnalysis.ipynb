```python
# ======================================================
# Code 02 — Basic Descriptive Analysis (Persuade 2.0)
# ======================================================
# Purpose:
#   Load the cleaned dataset and produce simple descriptive
#   summaries to set up later analyses.
#
# Input:
#   ./data/persuade_clean.csv (from Code 01)
# ======================================================

import pandas as pd
from IPython.display import display

# -------------------------------
# Load Data
# -------------------------------
df = pd.read_csv("./data/persuade_clean.csv")

# -------------------------------
# Dataset Overview
# -------------------------------
print("=== Dataset Info ===")
df.info()

# -------------------------------
# Gender Distribution
# -------------------------------
print("\n=== Gender Distribution ===")
gender_counts = df["gender"].value_counts(dropna=False).rename("count")
gender_perc = (
    df["gender"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
    .rename("percent")
)
gender_table = pd.concat([gender_counts, gender_perc], axis=1)
display(gender_table)

# -------------------------------
# Original Race/Ethnicity Distribution
# -------------------------------
print("\n=== Race/Ethnicity Distribution (Original) ===")
race_counts = (
    df["race_ethnicity"]
    .fillna("Missing")
    .value_counts(dropna=False)
    .rename("count")
)
race_perc = (
    df["race_ethnicity"]
    .fillna("Missing")
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
    .rename("percent")
)
race_table = pd.concat([race_counts, race_perc], axis=1)
display(race_table)

# -------------------------------
# Recode Race/Ethnicity
# White vs. non-White
# -------------------------------
df["race_binary"] = df["race_ethnicity"].apply(
    lambda x: "White"
    if pd.notna(x) and str(x).strip() == "White"
    else ("non-White" if pd.notna(x) else pd.NA)
)

# -------------------------------
# Recoded Race Distribution
# White vs. non-White
# -------------------------------
print("\n=== Race Distribution (White vs. non-White) ===")
race_bin_counts = df["race_binary"].value_counts(dropna=False).rename("count")
race_bin_perc = (
    df["race_binary"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
    .rename("percent")
)
race_bin_table = pd.concat([race_bin_counts, race_bin_perc], axis=1)
display(race_bin_table)

# -------------------------------
# Holistic Essay Score Distribution
# -------------------------------
print("\n=== Holistic Essay Score (1–6) Distribution ===")
score_counts = df["holistic_essay_score"].value_counts().sort_index().rename("count")
score_perc = (
    df["holistic_essay_score"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
    .rename("percent")
)
score_table = pd.concat([score_counts, score_perc], axis=1)
display(score_table)

print("\n=== Holistic Essay Score (1–3) Distribution ===")
score3_counts = df["holistic_essay_score_3cat"].value_counts().sort_index().rename("count")
score3_perc = (
    df["holistic_essay_score_3cat"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
    .rename("percent")
)
score3_table = pd.concat([score3_counts, score3_perc], axis=1)
display(score3_table)

# -------------------------------
# Holistic Essay Score by Gender
# Count + column percentage
# -------------------------------
print("\n=== Holistic Essay Score (1–6) by Gender ===")
gender_score_count = pd.crosstab(
    df["holistic_essay_score"],
    df["gender"],
    dropna=False
)
gender_score_pct = pd.crosstab(
    df["holistic_essay_score"],
    df["gender"],
    normalize="columns",
    dropna=False
).mul(100).round(2)

gender_score_table = gender_score_count.astype(str)
for col in gender_score_table.columns:
    gender_score_table[col] = (
        gender_score_count[col].astype(str)
        + " ("
        + gender_score_pct[col].astype(str)
        + "%)"
    )
display(gender_score_table)

print("\n=== Holistic Essay Score (1–3) by Gender ===")
gender_score3_count = pd.crosstab(
    df["holistic_essay_score_3cat"],
    df["gender"],
    dropna=False
)
gender_score3_pct = pd.crosstab(
    df["holistic_essay_score_3cat"],
    df["gender"],
    normalize="columns",
    dropna=False
).mul(100).round(2)

gender_score3_table = gender_score3_count.astype(str)
for col in gender_score3_table.columns:
    gender_score3_table[col] = (
        gender_score3_count[col].astype(str)
        + " ("
        + gender_score3_pct[col].astype(str)
        + "%)"
    )
display(gender_score3_table)

# -------------------------------
# Holistic Essay Score by Race
# Count + column percentage
# -------------------------------
print("\n=== Holistic Essay Score (1–6) by Race (White vs. non-White) ===")
race_score_count = pd.crosstab(
    df["holistic_essay_score"],
    df["race_binary"],
    dropna=False
)
race_score_pct = pd.crosstab(
    df["holistic_essay_score"],
    df["race_binary"],
    normalize="columns",
    dropna=False
).mul(100).round(2)

race_score_table = race_score_count.astype(str)
for col in race_score_table.columns:
    race_score_table[col] = (
        race_score_count[col].astype(str)
        + " ("
        + race_score_pct[col].astype(str)
        + "%)"
    )
display(race_score_table)

print("\n=== Holistic Essay Score (1–3) by Race (White vs. non-White) ===")
race_score3_count = pd.crosstab(
    df["holistic_essay_score_3cat"],
    df["race_binary"],
    dropna=False
)
race_score3_pct = pd.crosstab(
    df["holistic_essay_score_3cat"],
    df["race_binary"],
    normalize="columns",
    dropna=False
).mul(100).round(2)

race_score3_table = race_score3_count.astype(str)
for col in race_score3_table.columns:
    race_score3_table[col] = (
        race_score3_count[col].astype(str)
        + " ("
        + race_score3_pct[col].astype(str)
        + "%)"
    )
display(race_score3_table)

# -------------------------------
# Basic Text Length Statistics
# -------------------------------
print("\n=== Text Length Statistics (from full_text) ===")
full_text = df["full_text"].fillna("").astype(str)
char_count = full_text.str.len()
word_count = full_text.str.split().str.len()

summary = pd.DataFrame({
    "char_count": char_count.describe().round(2),
    "word_count": word_count.describe().round(2)
})
display(summary)
```
