```python
# ======================================================
# Code 02 — Basic Descriptive Analysis (Persuade 2.0)
# ======================================================
# Purpose:
#   Load the cleaned dataset and produce simple descriptive
#   summaries to set up later analyses.
#
# Input:
#   ./data/persuade_clean.csv (from Code 01)
# ======================================================

import pandas as pd
from IPython.display import display

# -------------------------------
# 1) Load Data
# -------------------------------
df = pd.read_csv("./data/persuade_clean.csv")

# -------------------------------
# 2) Dataset Overview
# -------------------------------
print("=== Dataset Info ===")
df.info()

# -------------------------------
# 3) Gender Distribution
# -------------------------------
print("\n=== Gender Distribution ===")
if "gender" in df.columns:
    counts = df["gender"].value_counts(dropna=False).rename("count")
    perc = df["gender"].value_counts(normalize=True, dropna=False).mul(100).round(2).rename("percent")
    gender_table = pd.concat([counts, perc], axis=1)
    display(gender_table)
else:
    print("Column 'gender' not found.")

# -------------------------------
# 4) Holistic Essay Score Distribution
# -------------------------------
print("\n=== Holistic Essay Score (1–6) Distribution ===")
if "holistic_essay_score" in df.columns:
    score_counts = df["holistic_essay_score"].value_counts().sort_index().rename("count")
    score_perc = df["holistic_essay_score"].value_counts(normalize=True).sort_index().mul(100).round(2).rename("percent")
    score_table = pd.concat([score_counts, score_perc], axis=1)
    display(score_table)
else:
    print("Column 'holistic_essay_score' not found.")


print("\n=== Holistic Essay Score (1–3) Distribution ===")
if "holistic_essay_score_3cat" in df.columns:
    score_counts = df["holistic_essay_score_3cat"].value_counts().sort_index().rename("count")
    score_perc = df["holistic_essay_score_3cat"].value_counts(normalize=True).sort_index().mul(100).round(2).rename("percent")
    score_table = pd.concat([score_counts, score_perc], axis=1)
    display(score_table)
else:
    print("Column 'holistic_essay_score_3cat' not found.")

# -------------------------------
# 5) Basic Text Length Statistics
# -------------------------------
print("\n=== Text Length Statistics (from full_text) ===")
if "full_text" in df.columns:
    full_text = df["full_text"].fillna("").astype(str)
    char_count = full_text.str.len()
    word_count = full_text.str.split().str.len()

    summary = pd.DataFrame({
        "char_count": char_count.describe().round(2),
        "word_count": word_count.describe().round(2)
    })
    display(summary)
else:
    print("Column 'full_text' not found.")
```
